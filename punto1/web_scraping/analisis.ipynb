# Análisis - Web Scraping Books to Scrape

```python
import pandas as pd

categorias = pd.read_parquet('categorias.parquet')
libros = pd.read_parquet('libros.parquet')
```

## 1. ¿Cuántas categorías de libros existen?

```python
n_categorias = categorias['categoria'].nunique()
print(n_categorias)
```

## 2. ¿Cuántos libros hay en cada categoría?

```python
libros_por_categoria = libros.groupby('categoria').size().sort_values(ascending=False)
libros_por_categoria
```

## 3. ¿Cuál es el libro más caro? (se muestran todos los empates)

```python
precio_max = libros['precio_con_impuesto'].max()
libros[libros['precio_con_impuesto'] == precio_max][['titulo', 'categoria', 'precio_con_impuesto']]
```

## 4. ¿Hay algún libro que aparezca en más de una categoría? (comparado por upc)

```python
conteo_upc = libros.groupby('upc')['categoria'].nunique()
upcs_repetidos = conteo_upc[conteo_upc > 1]
print(f"Libros (upc) que aparecen en más de una categoría: {len(upcs_repetidos)}")
libros[libros['upc'].isin(upcs_repetidos.index)][['upc', 'titulo', 'categoria']].sort_values('upc')
```

## 5. ¿Cuál es el libro más barato de cada categoría? (se muestran todos los empates)

```python
min_por_cat = libros.groupby('categoria')['precio_con_impuesto'].transform('min')
libros_baratos = libros[libros['precio_con_impuesto'] == min_por_cat]
libros_baratos[['categoria', 'titulo', 'precio_con_impuesto']].sort_values(['categoria', 'titulo'])
```

## 6. Diferencia de cada libro respecto al precio promedio de su categoría\n\ndiferencia = precio del libro - precio promedio de su categoría

```python
promedio_categoria = libros.groupby('categoria')['precio_con_impuesto'].transform('mean')
libros['diferencia_promedio_categoria'] = libros['precio_con_impuesto'] - promedio_categoria
libros[['titulo', 'categoria', 'precio_con_impuesto', 'diferencia_promedio_categoria']]
```

## 7. Libro con mayor ingreso potencial por categoría\n\ningreso potencial = precio_con_impuesto x cantidad_stock

```python
libros['ingreso_potencial'] = libros['precio_con_impuesto'] * libros['cantidad_stock']
ingreso_max_por_cat = libros.groupby('categoria')['ingreso_potencial'].transform('max')
libros_top_ingreso = libros[libros['ingreso_potencial'] == ingreso_max_por_cat]
libros_top_ingreso[['categoria', 'titulo', 'ingreso_potencial']].sort_values('categoria')
```
